```python
"""
Sistema de gestión de acceso tecnológico - Proyecto SIG-Conecta.
Este programa permite registrar el acceso de usuarios a un laboratorio de cómputo,
validando la información de entrada, consultando el último registro en memoria
y controlando el flujo mediante un menú interactivo.
"""

# Declaración de variables globales iniciales para almacenar el último registro
ultimo_id = 0          # Almacena el ID del usuario (entero)
ultimo_nombre = ""     # Almacena el nombre del usuario (cadena de texto)
ultima_pc = 0          # Almacena el número de computadora asignada (entero)
ultimo_nivel = ""      # Almacena el nivel de estudios del usuario (cadena de texto)

# Estructura de control de repetición (while) para el menú principal interactivo
while True:
    # Impresión de las opciones del menú principal en pantalla
    print("\n--- Menú Principal SIG-Conecta ---")
    print("1. Registrar ingreso")
    print("2. Consultar registro")
    print("3. Salir del sistema")

    # Captura de la opción seleccionada por el administrador
    opcion = input("Seleccione una opción (1-3): ")

    # Estructura de control selectiva (if-elif-else) para evaluar la opción ingresada
    if opcion == '1':
        # Variable para controlar el ciclo de continuidad de registros
        continuar = 'S'

        # Bucle para permitir múltiples registros sin salir al menú principal
        while continuar.upper() == 'S':
            print("\n--- Registrar Ingreso ---")

            # Bucle de validación para el ID del usuario
            while True:
                id_input = input("Ingrese el ID de usuario (1-1000): ")
                # Validar que el dato sea numérico y esté en el rango de 1 a 1000
                if id_input.isdigit() and 1 <= int(id_input) <= 1000:
                    ultimo_id = int(id_input) # Se guarda el valor válido
                    break                     # Se rompe el bucle de validación y avanza
                else:
                    # Notifica el error si el ID está fuera de rango o contiene letras
                    print("Error: El ID debe ser un número entero positivo entre 1 y 1000. Intente de nuevo.")

            # Captura del nombre del usuario (no requiere validación numérica)
            ultimo_nombre = input("Ingrese el nombre del usuario: ")

            # Bucle de validación para el número de PC asignada
            while True:
                pc_input = input("Ingrese el número de PC asignada (1-20): ")
                # Validar que el dato sea numérico y esté en el rango de 1 a 20
                if pc_input.isdigit() and 1 <= int(pc_input) <= 20:
                    ultima_pc = int(pc_input) # Se guarda el valor válido
                    break                     # Se rompe el bucle de validación y avanza
                else:
                    # Notifica el error si el número de PC es inválido
                    print("Error: El número de PC debe estar entre 1 y 20. Intente de nuevo.")

            # Captura del nivel de estudios del usuario
            ultimo_nivel = input("Ingrese el nivel de estudios (ej. Primaria, Secundaria): ")

            # Muestra un mensaje de confirmación con los datos recién capturados
            print("\n¡Datos capturados exitosamente!")
            print(f"ID: {ultimo_id}, Nombre: {ultimo_nombre}, PC: {ultima_pc}, Nivel: {ultimo_nivel}")

            # Ciclo de continuidad: pregunta si se desea registrar otro usuario
            continuar = input("\n¿Desea registrar otro acceso? (S/N): ")
            # Si el usuario ingresa algo distinto a 'S' o 's', el bucle while interno termina y regresa al menú principal

    elif opcion == '2':
        # Opción para consultar el último registro guardado en memoria
        print("\n--- Consultar Último Registro ---")
        # Verifica si ya se ha registrado al menos un usuario (ID diferente de 0)
        if ultimo_id == 0:
            print("Aún no hay usuarios registrados en el sistema.")
        else:
            # Imprime los valores almacenados en las variables globales
            print(f"ID del usuario: {ultimo_id}")
            print(f"Nombre del usuario: {ultimo_nombre}")
            print(f"PC asignada: {ultima_pc}")
            print(f"Nivel de estudios: {ultimo_nivel}")

    elif opcion == '3':
        # Opción para salir del sistema
        print("\nSaliendo del sistema SIG-Conecta. ¡Hasta pronto!")
        break # Rompe el ciclo while principal, finalizando así la ejecución del programa

    else:
        # Control de error: notifica si el usuario ingresó una opción distinta a 1, 2 o 3 en el menú principal
        print("\nError: Opción inválida. Por favor, ingrese 1, 2 o 3.")
```
